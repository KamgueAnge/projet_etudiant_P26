## Commentaires de l'APP

Pourquoi quand j'utilise 4 actifs dans le portefeuille, il y a 1 ou 2 qui descendent à 0?

C’est la grande magie de l'optimisation selon Harry Markowitz (et le critère du maximum de Sharpe) ! Ce comportement n'est pas un bug du code, c'est un résultat financier tout à fait normal et ultra-classique.

Lorsque l'algorithme met à 0% certains actifs (comme Total ou BNP Paribas dans tes essais précédents), il fait son travail de sélection drastique. Voici exactement pourquoi cela se produit :

1. La recherche du Ratio de Sharpe Maximum 🏆
L'objectif de ton application est de trouver le portefeuille qui offre le maximum de rendement pour chaque unité de risque (volatilité) prise.
Si l'actif A (ex: LVMH) a eu une croissance exceptionnelle avec une volatilité modérée, et que l'actif B (ex: BNP Paribas) a stagné sur la même période tout en étant très volatil, l'algorithme va mathématiquement refuser d'acheter l'actif B. Ajouter l'actif B dégraderait le Ratio de Sharpe global du portefeuille. L'algorithme lui attribue donc une allocation de 0 %.

2. La contrainte "Long-Only" (Pas de vente à découvert) 🚫
Dans ton code, nous avons défini la contrainte suivante : weight_bounds=(0, 1).

Cela signifie qu'il est interdit d'avoir des poids négatifs (pas de vente à découvert ou short selling).

Si cette contrainte n'existait pas, l'algorithme proposerait probablement de "shorter" (vendre) les actifs moins performants (poids de −20%) pour investir encore plus sur les meilleurs (poids de 120%).

Comme le seuil minimum est bloqué à 0, l'algorithme se contente de les exclure totalement en les calant à 0,00 %.

3. Le phénomène de concentration de Markowitz 📉
La théorie moderne du portefeuille cherche à diversifier pour réduire le risque, mais seulement si les actifs se complètent bien (faible corrélation).
Si deux actifs sont très corrélés (ils montent et descendent en même temps) mais que l'un des deux surperforme largement l'autre, l'optimiseur va couper l'actif le moins performant pour concentrer le capital sur le meilleur. C'est pour cela qu'on se retrouve très souvent avec des portefeuilles concentrés sur seulement 2 ou 3 actifs majeurs, tandis que les autres tombent à zéro.

1. Le Biais d'Anticipation (Le plus probable)
Dans ton calcul de mean_historical_return et de la matrice de covariance S, tu utilises tout l'historique pour calculer les poids.

Ton modèle dit : "En 2025, j'achète les actions qui ont été les meilleures entre 2010 et 2025."

Le problème : C'est tricher ! Tu as utilisé les rendements de 2025 pour décider ce que tu devais acheter... en 2025.

La réalité : Le marché ne te donne pas le rendement de l'année avant qu'elle ne soit finie. Ton modèle est donc "parfait" car il choisit les gagnants après les avoir vus gagner.

2. Le Biais de Survie (Liste Wikipédia)
Tu as récupéré la liste des entreprises du CAC 40 actuelles sur Wikipédia.

Le problème : Ces entreprises sont les "survivantes". Beaucoup d'entreprises qui faisaient partie du CAC 40 en 2010 ont fait faillite, ont été rachetées ou ont été sorties de l'indice parce qu'elles allaient mal.

L'impact : En testant ton portefeuille uniquement sur les 40 meilleures entreprises d'aujourd'hui, tu ignores toutes les entreprises qui ont échoué durant les 15 dernières années. Ton univers d'actifs est donc artificiellement dopé par les succès actuels.

3. La Fréquence de Rééquilibrage
Ton modèle calcule une allocation "optimale" sur toute la période, mais en pratique, il ne rééquilibre jamais le portefeuille. Il calcule les poids une seule fois à la fin. Ton algorithme maintient mécaniquement les positions gagnantes pendant 15 ans, ce qu'aucun gestionnaire humain ne peut faire sans prendre de profits ou subir des chocs de liquidité.

🚀 Comment corriger cela pour avoir une valeur réelle (crédible) ?
Pour que ton jury soit impressionné, tu dois passer d'un "Backtest Statique" à un "Backtest glissant" (Rolling Backtest). Au lieu de calculer les poids une seule fois, tu dois les recalculer à intervalle régulier (par exemple tous les mois) en n'utilisant que les données passées.

Voici la modification logique à apporter à ton rapport/soutenance :

Explique la différence : Présente le résultat actuel comme une "Limite supérieure théorique" (la performance maximale atteignable si on avait une boule de cristal).

Propose l'approche glissante : Explique que pour obtenir la valeur réelle, il faut effectuer un rééquilibrage mensuel. Le modèle ne doit utiliser que les données de t-12 mois pour prédire le poids de t.

Ce modèle est une optimisation théorique qui expose les limites du marché. En pratique, un rééquilibrage dynamique (glissant) réduirait légèrement la performance du fait des coûts de transaction, mais ma version actuelle met en lumière le potentiel d'optimisation pur du portefeuille selon les principes de Markowitz.
